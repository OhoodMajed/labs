# Topic 1 : Predicting Vehicle Accident Severity

## Problem Statement

Traffic accidents cause thousands of injuries and major traffic disruptions every year across the US. Emergency responders, insurers, and city planners currently have limited tools to anticipate how severe an accident is likely to be given only the conditions surrounding it (location, weather, road features, time of day). This project asks: **can we predict the severity of a traffic accident from the conditions present at the time it occurred?**

## Goals

- Build a model that classifies an accident's severity level (1-4) from environmental and road-related features.
- Identify which weather, road, and time-of-day conditions are most associated with severe accidents.
- Provide insights that could help city planners prioritize infrastructure changes (signals, signage, lighting) at high-risk locations.
- Provide insights that could help emergency services better anticipate resource needs based on conditions.

## Approach

1. Clean and subset the raw data (7M+ rows) to a manageable, representative sample.
2. Engineer time-based features (hour, day of week, rush hour flag) from timestamp fields.
3. Explore correlations between severity and weather/road-feature columns.
4. Handle missing values in weather columns (temperature, visibility, wind speed) via imputation.
5. Train baseline classification models (Logistic Regression, Random Forest) predicting `Severity`.
6. Evaluate with metrics appropriate for an imbalanced multi-class target (macro F1, confusion matrix).

## Data Info
Data Source : [US Accidents (2016 - 2023), Kaggle](https://www.kaggle.com/datasets/sobhanmoosavi/us-accidents) — ~7.7 million rows, 46 columns

## Data Dictionary
| Column    | Type    | Description |
| --------  | ------- | -------     |
| ID | object | Unique identifier for the accident record |
| Severity | int | Accident severity, 1 (least) to 4 (most) — **target column** |
| Start_Time | datetime | Start time of the accident |
| End_Time | datetime | End time (when impact on traffic flow was cleared) |
| Start_Lat / Start_Lng | float | GPS coordinates of the accident start point |
| Distance(mi) | float | Length of the road affected by the accident |
| City / State | object | Location of the accident |
| Temperature(F) | float | Temperature at the time of the accident |
| Humidity(%) | float | Humidity at the time of the accident |
| Visibility(mi) | float | Visibility distance |
| Weather_Condition | object | Weather description (Rain, Fog, Clear, etc.) |
| Sunrise_Sunset | object | Whether it was day or night |
| Crossing / Junction / Traffic_Signal | bool | Presence of nearby road features |

# Topic 2 : Credit Card Fraud Detection

## Problem Statement

Credit card fraud costs banks and consumers billions of dollars annually, and fraudulent transactions are extremely rare compared to legitimate ones, which makes them hard to catch with simple rules. This project asks: **can we accurately flag fraudulent transactions among millions of legitimate ones, despite the severe class imbalance?**

## Goals

- Build a classifier that flags a transaction as fraudulent or legitimate.
- Maximize recall on fraud cases without producing an unmanageable number of false positives.
- Understand which anonymized features are most predictive of fraud.
- Propose a strategy (resampling, class weighting, threshold tuning) suited to highly imbalanced data.

## Approach

1. Load and inspect the transaction data; confirm the severe class imbalance (fraud is <1% of rows).
2. Scale the `Amount` and `Time` columns (the V1-V28 features are already PCA-transformed/scaled).
3. Visualize class distribution and correlations between PCA components and the fraud label.
4. Address imbalance using techniques such as SMOTE, undersampling, or class-weighted models.
5. Train and compare models (Logistic Regression, Random Forest, XGBoost).
6. Evaluate using precision, recall, F1, and ROC-AUC — **not** plain accuracy.

## Data Info
Data Source : [Credit Card Fraud Dataset, Kaggle](https://www.kaggle.com/datasets/ahmedjawed/creditscardfrauddataset) — ~284,807 rows, 31 columns

## Data Dictionary
| Column    | Type    | Description |
| --------  | ------- | -------     |
| Time | int | Seconds elapsed between this transaction and the first transaction in the dataset |
| V1 ... V28 | float | Anonymized features obtained via PCA transformation (original features withheld for privacy) |
| Amount | float | Transaction amount |
| Class | int | 1 = fraudulent transaction, 0 = legitimate — **target column** |

# Topic 3 : Food & Nutrition Health Classification

## Problem Statement

Consumers and app developers increasingly want quick, trustworthy signals about how healthy a packaged or whole food is, but nutrition labels are dense and hard to compare across products. This project asks: **can we predict a food's health/Nutri-Score grade from its raw nutritional composition, and which nutrients drive that classification?**

## Goals

- Build a model that predicts a food's Nutri-Score grade (A-E) or health-score tier from macro/micronutrient values.
- Identify which nutrients (sugar, sodium, fiber, saturated fat, etc.) most influence the health classification.
- Compare healthiness patterns across food categories (fruits, snacks, beverages, dairy, etc.).
- Explore how allergen presence correlates with processing level (NOVA classification).

## Approach

1. Load the combined USDA / Open Food Facts nutrition data and inspect category balance.
2. Clean and standardize nutrient units (per 100g) and handle missing values common in crowd-sourced data.
3. Explore distribution of the target (Nutri-Score grade) and check for class imbalance across grades.
4. Examine correlations between nutrients (sugar, sodium, saturated fat, fiber) and the target grade.
5. Train classification models (Random Forest, Gradient Boosting) to predict the grade.
6. Evaluate with macro F1 / accuracy and inspect feature importance for interpretability.

## Data Info
Data Source : [Global Food & Nutrition Database 2026, Kaggle](https://www.kaggle.com/datasets/ahsanneural/global-food-and-nutrition-database-2026) — 45,000+ foods, from USDA FoodData Central & Open Food Facts

## Data Dictionary

| protein_g / fat_g / carbs_g | float | Macronutrient content per 100g |
| sugar_g / fiber_g / sodium_mg | float | Additional nutrient content per 100g |

# Topic 4 : Customer Churn Prediction (Telecom)

## Problem Statement

Telecom providers lose significant recurring revenue when customers cancel their service, and acquiring a new customer is far more expensive than retaining an existing one. This project asks: **can we predict which customers are likely to churn based on their account, billing, and usage patterns, in time for the business to intervene?**

## Goals

- Build a model that predicts the probability a customer will churn.
- Identify the strongest drivers of churn (contract type, billing, tenure, support interactions).
- Segment customers into risk tiers to help prioritize retention campaigns.
- Estimate the potential revenue impact of reducing churn by a given percentage.

## Approach

1. Load the 1M-row synthetic telecom dataset and check the overall churn rate (target balance).
2. Explore churn rate broken out by contract type, tenure, and payment method.
3. Engineer features such as tenure buckets and average monthly spend.
4. Check for multicollinearity among billing-related features.
5. Train classification models (Logistic Regression, Random Forest) predicting `churn`.
6. Evaluate with ROC-AUC and precision/recall, since churners are the minority class.

## Data Info
Data Source : [Customer Churn Prediction Dataset 1M, Kaggle](https://www.kaggle.com/datasets/isandeep06/customer-churn-prediction-dataset-1m) — 1,000,000 rows, 28 columns (synthetic telecom data)
