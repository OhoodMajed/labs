## 0. Imports & Load Data

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Read the data
# NOTE: update the path/filename below to match your downloaded file exactly
df = pd.read_csv("US_Accidents.csv")

print("File loaded successfully ✅")
```

```
# Shape of the raw data (before any cleaning)
print("Rows and Columns (before cleaning):")
print(df.shape)
```

```
# Preview first 5 rows
df.head()
```

```
df.info()
```

## 3. Missing Values
This is important — the presentation requires: *"How many missing values? What is your plan?"*

```
missing = df.isnull().sum()
missing_percentage = (missing / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": missing_percentage
})

missing_table.sort_values(by="Missing Values", ascending=False).head(20)
```

**Plan for missing values** *(fill in once you see the actual output above)*:
- Columns with very high missing % (e.g. >40-50%) → candidates to drop
- Columns with low missing % → impute (median/mode) or drop rows depending on importance
- Document your reasoning per column here.

## 4. Duplicate Rows

```
print("Duplicate rows:", df.duplicated().sum())
```

```
# Drop duplicates if any were found
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

## 5. Column Integrity Check

```
original_columns = df.columns
print("Original column count:", len(original_columns))
```

## 6. Target Column Check — `Severity`
`Severity` is our target for this project (accident severity prediction).

```
df['Severity'].value_counts()
```

```
plt.figure(figsize=(6,4))
sns.countplot(x='Severity', data=df)
plt.title("Accident Severity Distribution")
plt.xlabel("Severity")
plt.ylabel("Count")
plt.show()
```

**Target type:** Severity is discrete/ordinal (categorical) with a small number of classes → this is a **classification** problem.
Check the plot above for class balance — accident severity datasets are often imbalanced toward the middle classes (e.g. 2 and 3).

## 7. Feature Types — Numerical vs Categorical
Answers the requirement: *"How many categorical and continuous columns do you have?"*

```
numeric_cols = df.select_dtypes(include=np.number).columns
categorical_cols = df.select_dtypes(include='object').columns

print("Numerical columns:", len(numeric_cols))
print("Categorical columns:", len(categorical_cols))

print("\nNumerical:", list(numeric_cols))
print("\nCategorical:", list(categorical_cols))
```
